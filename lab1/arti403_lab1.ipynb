## Task #2 — Loading and Visualizing Images

> **Before running the cells below:** create a folder named `images` next to this notebook and place the two files below inside it. These are standard grayscale test images used throughout image-processing courses — ask your instructor for them if you do not already have them from the course resources.
> * `images/cameraman.tif`
> * `images/lena_gray_256.tif`

### a. Loading/reading an image using OpenCV

`cv2.imread` loads the image file into a NumPy array, which is then displayed with `plt.imshow`.

```python
img = cv2.imread('images/cameraman.tif')  # Load the image data into a NumPy array

plt.imshow(img)
plt.show()
```

### b. Loading/reading image using PIL

`Image.open` loads the file into a PIL `Image` object. Since the image is grayscale, it is displayed with the `Greys_r` colormap so Matplotlib does not tint it.

```python
img2 = Image.open('images/lena_gray_256.tif')  # Load image into a PIL Image object

plt.imshow(img2, cmap=cm.Greys_r)
plt.show()
```

---
## Task #3 — Image Storing

`img` (loaded with OpenCV) and `img2` (loaded with PIL) from Task #2 can each be written back to disk.

### a. Saving an image to disk using OpenCV

`cv2.imwrite` writes a NumPy image array to disk, choosing the format from the file extension.

```python
cv2.imwrite('new_image.jpg', img)
```

### b. Saving an image to disk using PIL

PIL images have their own `save` method, which also chooses the format from the file extension.

```python
img2.save('new_image2.jpg')
```

## Task #4 — Display Image as an Array

An image loaded with OpenCV is already a NumPy array. A PIL image can be converted into one with `np.array`.

### a. The OpenCV image as an array

`img.shape` gives its dimensions, and printing `img` shows the raw pixel values.

```python
print(img.shape)  # Note: img loaded using OpenCV is a NumPy array
print(img)
```

### b. Converting the PIL image to a NumPy array

`np.array()` converts the PIL image into a NumPy array so it can be inspected the same way.

```python
# Convert the image to a NumPy array
img_array = np.array(img2)

# Print the array
print(img_array.shape)
print(img_array)
```
